```python
import pandas as pd
import numpy as np
```

```python
df_wm = pd.read_csv('/home/mike/Documents/Economic Data Science/Case Study/Data/wohnungsmieten.csv', delimiter=',', encoding='UTF-8', low_memory=False)
```

```python
df_wm
```

```python
df_wm = df_wm.replace(r"^\s*$", np.nan, regex=True)
```

```python
def parse_year_month(series):
    s = series.astype(str).str.strip()
    s = s.replace({'nan': None, 'None': None})
    s = s.where(s.isna(), s.str.replace(r'(?i)m', '-', regex=True))
    dt = pd.to_datetime(s, format='%Y-%m', errors='coerce')
    return dt
```

```python
df_wm['adat_dt'] = parse_year_month(df_wm['adat'])
df_wm['edat_dt'] = parse_year_month(df_wm['edat'])

df_wm['adat_mm_yyyy'] = df_wm['adat_dt'].dt.strftime('%m.%Y')
df_wm['edat_mm_yyyy'] = df_wm['edat_dt'].dt.strftime('%m.%Y')
```

```python
df_wm["baujahr"] = pd.to_numeric(df_wm["baujahr"], errors="coerce")
df_wm["wohnflaeche"] = pd.to_numeric(df_wm["wohnflaeche"], errors="coerce")
df_wm["nutzflaeche"] = pd.to_numeric(df_wm["nutzflaeche"], errors="coerce")
df_wm["mietekalt"] = pd.to_numeric(df_wm["mietekalt"], errors="coerce")
df_wm["nebenkosten"] = pd.to_numeric(df_wm["nebenkosten"], errors="coerce")
df_wm["heizkosten"] = pd.to_numeric(df_wm["heizkosten"], errors="coerce")
df_wm["rent_sqm"] = pd.to_numeric(df_wm["rent_sqm"], errors="coerce")
```

```python
df_wm["baujahr"] = pd.to_numeric(df_wm["baujahr"], errors="coerce").round(0).astype("Int64")
df_wm["plz"] = pd.to_numeric(df_wm["plz"], errors="coerce").round(0).astype("Int64")
```

```python
mask_wohnflaeche = df_wm["wohnflaeche"].notna()
df_wm.loc[mask_wohnflaeche, "wohnflaeche"] = df_wm.loc[mask_wohnflaeche, "wohnflaeche"].round(2)

mask_nutzflaeche = df_wm["nutzflaeche"].notna()
df_wm.loc[mask_nutzflaeche, "nutzflaeche"] = df_wm.loc[mask_nutzflaeche, "nutzflaeche"].round(2)

mask_mietkalt = df_wm["mietekalt"].notna()
df_wm.loc[mask_mietkalt, "mietekalt"] = df_wm.loc[mask_mietkalt, "mietekalt"].round(2)

mask_nebenkosten = df_wm["nebenkosten"].notna()
df_wm.loc[mask_nebenkosten, "nebenkosten"] = df_wm.loc[mask_nebenkosten, "nebenkosten"].round(2)

mask_heizkosten = df_wm["heizkosten"].notna()
df_wm.loc[mask_heizkosten, "heizkosten"] = df_wm.loc[mask_heizkosten, "heizkosten"].round(2)

mask_rent_sqm = df_wm["rent_sqm"].notna()
df_wm.loc[mask_rent_sqm, "rent_sqm"] = df_wm.loc[mask_rent_sqm, "rent_sqm"].round(2)

```
